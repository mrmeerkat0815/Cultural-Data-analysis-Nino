# Task 2: Mini research project (group, 90 pts)
For the main group part, we have combined the lessons and methods from the entire course into one (not so) mini project. You have to fill out the notebook below, write a 2.000-word report, and a 5-8 minute video.

### Notebook (30pts)
This **notebook** is meant to show how you carried out your mini-research. It includes code on data collection, analysis, and visualization. What data and method you choose is up to you, but they need to contain the following components:

- **Social media data collection:** Your project should (at least partially) analyse traces from social media. This is for instance possible through interfacing with an API, or by loading in data from external tools or archives. Think, for instance, about GUI tools like the YouTube Data Tools, Zeeschuimer, 4CAT, or [Kenniskrabber](https://github.com/digitalmethodsinitiative/kenniskrabber), or libraries that you can use in the code here like `yt-dlp`. Make sure that you don't include any API keys or personal data, for example by deleting or anonymizing usernames.
- **Statistical analysis**: The notebook should include (preferably noteworthy) statistics on the data collected.
- **Text mining / NLP**: The notebook needs to contain an analysis of the textual contents in your social media dataset. The analysis must include *at least 2 traditional methods*, like topic modelling, co-words, tf-idf, etc. Also conduct *an analysis using a model from Hugging Face* to study the features of your text. You can think here about classification models, embedding models or even use a small LLM to produce some insights.
- **Image analysis**: You should analyse images that are related to your social media case. The images may be derived directly from your social media dataset (e.g. YouTube thumbnails) but they can also come from external archives. The images should then be analysed with *a traditional method* like colour analysis and *through a model imported from Hugging Face*.

For each part, try to think of original and well-fitting methods. Include informative visualisations and text cells that briefly substantiate your process and decisions. We added a template in the cells below, but how and in what order you carry out these analyses is up to you--try to think of a logical process.

### Report (30pts)
With the notebook as source material, write a researchreport that (1) introduces your research interest and question(s), (2) connects them to the academic literature (also from the MA course), (3) presents and interprets your main findings, and (4) critically reflects on your methods and their limits. This is less about the individual pieces of code and more about a coherent research argument interpreting the findings and situating the project. What have you found about your case? Were you able to answer your initial question(s)? What did the methods allow you to observe, and where are their limitations?

### Video (30pts)
Record a 5 to 8-minute long video on your project. The video should briefly introduce the case and findings, but is main purpose is to step back and reflect on the **process** of the research, including the use of generative AI. Example questions can be: How did you divide the work? How did you come up with your questions? What were the main struggles? How did you use generative AI in the process? If you did use it, how did it affect your own thinking and the decision-making? Did the LLM nudge you in certain directions rather than others?

Use examples from your research process to illustrate your reflections. Keep the presentation of your case and findings brief, and focus on your decisions, challenges, and learning process.

A simple montage with a voice-over or video overlay will do; it's about the content of the video rather than its aesthetics.

Include the video as .mp4 in your assignment on Canvas or as a link to YouTube or GitHub.

### Grading criteria
We will grade for originality, critical engagement, cohesiveness, comprehensiveness, academic embeddedness, the choice and justification of methods and models, and quality of analyses, reflections, code, video, and writing. The video should be well-presented and followable, but we will not grade on fancy editing or visuals.
