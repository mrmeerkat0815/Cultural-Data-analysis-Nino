## Real-life Dataset Exercise

As a final exercise, we use our knowledge to analyse a famous real-life social dataset about death penalties. We will work through a detailed example of data exploration and analyse a very small part of the complex question of racism in the US legal system.

The data `deathpenalty_array` covers judgments of defendants in cases of multiple murders in Florida between 1976 and 1987. The cases all have features that (a) describe whether the death penalty was handed out (where 0 refers to no, 1 to yes), (b) the race of the victim and (c) the race of the defendant (black is referred to as 0, white as 1).

> **Remember:** It is **very** problematic to record non-binary attributes like 'race' in this way. 'Race' is a highly contested idea, and we should be more careful than the data we use, which assigns people to two very simplified categories. Discussing such constraints should be part of your own critical data analysis. But you are only at the beginning of your programming journey, and we will ignore this for now to make things not too complicated – as well as to show you how datafication flattens complex realities.

Load the dataset by running the cell below. Take a moment to think about what you have just loaded. How is the data organised?

````python
#Keep cell

import numpy as np

deathpenalty_array = np.array([
[0,	0,	0,	139],
[1,	0,	0,	4],
[0,	1,	0,	37],
[1,	1,	0,	11],
[0,	0,	1,	16],
[1,	0,	1,	0],
[0,	1,	1,	414],
[1,	1,	1,	53]])

deathpenalty = deathpenalty_array[:,0]
victimrace = deathpenalty_array[:,1]
defendantrace = deathpenalty_array[:,2]
freq = deathpenalty_array[:,3]

print('Death Penalty :', deathpenalty)
print('Victim Race   :', victimrace)
print('Defendant Race:', defendantrace)
print('Frequency     :', freq)
````

> **Who decided?** The researchers decided what counts as a case: trials for multiple murders in Florida between 1976 and 1987. They recorded race in two categories, and they chose to compare judgments by the race of the defendant and of the victim. Every row is a combination of these decisions, and the column `freq` counts how many cases fall into it. Decisions taken before a trial, such as what the prosecution charged, are not in the data.

Remember that a black person is represented by 0 and a white by 1. A death penalty was handed out if the column `deathpenalty` equals 1, with 0 otherwise. This is quite a confusing way of recording events and not recommended beyond the use in such a small example. It was, however, quite common in those early days of data.

`deathpenalty_array` records all the information, and we use indexes to select particular columns of the data to define:

- The array `freq` for frequency of each combination, which has 674 cases in total.
- `deathpenalty`: Was the judgment a death penalty? yes = 1, no = 0.
- `victimrace`: the race of the victim; white = 1, black = 0.
- `defendantrace`: the race of the defendant; white = 1, black = 0.

With the NumPy function `np.mean()`, you can retrieve the average of a series of numbers. You can  answer what the average frequency of judgments is by entering ```np.mean(freq)```. Try it.

What we have just done is also called summary statistics with numerical data, which is basically anything measured in numbers.

> **Background: the data, and Simpson's paradox** (optional reading)
>
> The data comes from a study by Michael Radelet and Glenn Pierce, 'Choosing those who will die: race and the death penalty in Florida' (*Florida Law Review*, 1991). It has become a textbook example in statistics, for instance in Alan Agresti's *Categorical Data Analysis*, because it shows *Simpson's paradox* (https://en.wikipedia.org/wiki/Simpson%27s_paradox): a comparison can reverse when you split the data into groups.
>
> Overall, white defendants received the death penalty more often: 53 of 483 (11%), against 15 of 191 black defendants (8%). But split by the race of the victim, the picture reverses:
> - when the victim was white, 11 of 48 black defendants (23%) received the death penalty, against 53 of 467 white defendants (11%);
> - when the victim was black, 4 of 143 black defendants (3%) did, and none of the 16 white defendants.
>
> The reason is the victim's race. Killing a white person led to a death penalty far more often (64 of 515 cases, 12%) than killing a black person (4 of 159, 3%), and most defendants were tried for killing someone of their own race. The overall numbers hide this pattern. Before you trust a comparison, ask which other variable might be behind it.
>

> **Background: reading a boxplot** (optional reading)
>
> The box spans the middle 50% of the values: from the first quartile (25%) to the third quartile (75%), which you calculated with `np.quantile`. The line inside the box is the median. The whiskers reach out to the most extreme values that lie within 1.5 times the height of the box; values beyond are drawn as single dots, the outliers (https://en.wikipedia.org/wiki/Box_plot).
