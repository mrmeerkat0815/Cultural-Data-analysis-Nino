For the the main group part, we have combined the lessons and methods from the entire course into one (not so) mini project. You have to fill out the notebook below *and* write a 2.000-word report to reflect on your methodology and findings.

### Notebook (50pts)
This **notebook** is meant to show how you carried out your mini-research. It includes code on data collection, analysis, and visualization. What data and method you choose is up to you, but they need to contain the following components:

- **Social media data collection:** Your project should (at least partially) analyse traces from social media. This is for instance possible through the YouTube Data Tools or other tools like `youtube-dl`, Zeeschuimer, and 4CAT. You could also use APIs directly from this notebook (e.g. [BlueSky](https://docs.bsky.app/)). Make sure that you don't include any API keys or personal data, for example by deleting or anonymizing usernames.
- **Statistical analysis**: The report should include (preferably noteworthy) statistics on the data collected.
- **Text mining / NLP**: The notebook needs to contain an analysis of the textual contents in your social media dataset. The analysis must include *at least 2 traditional methods*, like topic modelling, co-words, tf-idf, etc. Also conduct *an analysis using a model from Hugging Face* to study the features of your text, like emotional tone, opinion or affective state.
- **Image analysis**: You should analyse images that are related to your social media case. The images may be derived directly from your social media dataset (e.g. YouTube thumbnails) but they can also come from external archives. The images should then be analysed with *a traditional method* like colour analysis and *through a model imported from Hugging Face*.

For each part, try to include informative visualisations and text cells that briefly substantiate your process and decisions. We added a template in the cells below, but how and in what order you carry out these analyses is up to you--try to think of a logical process.

### Report (30pts)
With the notebook as source material, write a **report** that 1) introduces your research interests and question(s), 2) summarizes your main findings, 3)
critically reflects on your methodology and its limitations and 4) reflects on your use of generative AI for this project. This is less about the individual pieces of code and more about taking a step back and contextualizing your project. You can ask: What have you found about your case? Were you able to answer your initial question(s)? What did the methods allow you to observe, and where are their limitations?
